## 2. Storage Service (`storageService.ts`)

Handles all data persistence operations using LocalStorage with simulated async delays.

### Key Methods


````
storage_service = '''export const storageService = {
  async getNotes(): Promise<Note[]> {
    await delay(DELAY_MS);
    return getNotesFromStorage();
  },

  async saveNote(note: Note): Promise<void> {
    const notes = getNotesFromStorage();
    const existingIndex = notes.findIndex(n => n.id === note.id);
    
    const updatedNote = {
      ...note,
      updatedAt: Date.now(),
      createdAt: existingIndex >= 0 ? notes[existingIndex].createdAt : Date.now()
    };
    
    if (existingIndex >= 0) {
      notes[existingIndex] = updatedNote;
    } else {
      notes.push(updatedNote);
    }
    
    saveNotesToStorage(notes);
  },

  async deleteNote(noteId: string): Promise<void> {
    const notes = getNotesFromStorage();
    const filtered = notes.filter(n => n.id !== noteId);
    saveNotesToStorage(filtered);
  },

  createEmptyNote(): Note {
    return {
      id: crypto.randomUUID(),
      title: '',
      content: '',
      status: NoteStatus.DRAFT,
      category: NoteCategory.UNCATEGORIZED,
      tags: [],
      totalTime: 0,
      sessions: [],
      createdAt: Date.now(),
      updatedAt: Date.now(),
    };
  }
};'''

Markdown(f"```typescript\n{storage_service}\n```")
````

## 6. Interactive Demo (Python Simulation)


````
import json
from datetime import datetime
import uuid

def create_note(title, content, status="Draft", category="Uncategorized"):
    return {
        "id": str(uuid.uuid4()),
        "title": title,
        "content": content,
        "status": status,
        "category": category,
        "tags": [],
        "totalTime": 0,
        "sessions": [],
        "createdAt": int(datetime.now().timestamp() * 1000),
        "updatedAt": int(datetime.now().timestamp() * 1000)
    }

sample_notes = [
    create_note("Q4 Planning Meeting", "Discussed roadmap for Q4. Key priorities: Launch feature X, Improve performance, Hire engineers.", "In Progress", "Work"),
    create_note("Book Ideas", "Potential book about AI and productivity.", "Draft", "Ideas"),
    create_note("Grocery List", "Milk, eggs, bread, vegetables, fruits", "Completed", "Personal")
]

print("Sample Notes Created:")
print("=" * 50)
for i, note in enumerate(sample_notes, 1):
    print(f"{i}. {note['title']} [{note['category']}] - {note['status']}")
````

````
def simulate_focus_session(note, duration_minutes):
    start_time = int(datetime.now().timestamp() * 1000)
    end_time = start_time + (duration_minutes * 60 * 1000)
    
    session = {
        "id": str(uuid.uuid4()),
        "startTime": start_time,
        "endTime": end_time,
        "duration": duration_minutes * 60
    }
    
    note["sessions"].append(session)
    note["totalTime"] += duration_minutes * 60
    
    return session

work_note = sample_notes[0]
for duration in [25, 45, 30]:
    simulate_focus_session(work_note, duration)

print(f"Total focus time on '{work_note['title']}': {work_note['totalTime'] / 60:.0f} minutes")
print(f"Sessions: {len(work_note['sessions'])}")
````

### Productivity Visualization


````
import pandas as pd

df = pd.DataFrame([{
    'title': note['title'],
    'category': note['category'],
    'status': note['status'],
    'total_minutes': note['totalTime'] / 60,
    'sessions': len(note['sessions'])
} for note in sample_notes])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

colors = {'Work': '#3B82F6', 'Personal': '#10B981', 'Ideas': '#F59E0B'}
df.groupby('category')['total_minutes'].sum().plot(
    kind='bar', ax=axes[0],
    color=[colors.get(cat, '#ccc') for cat in df.groupby('category')['total_minutes'].sum().index]
)
axes[0].set_title('Focus Time by Category', fontsize=14, fontweight='bold')
axes[0].set_ylabel('Minutes')

status_colors = {'Draft': '#9CA3AF', 'In Progress': '#F59E0B', 'Completed': '#10B981'}
df['status'].value_counts().plot(
    kind='pie', ax=axes[1],
    colors=[status_colors.get(s, '#ccc') for s in df['status'].value_counts().index],
    autopct='%1.0f%%'
)
axes[1].set_title('Note Status Distribution', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('/workspace/productivity_chart.png', dpi=150, bbox_inches='tight')
plt.show()
````
